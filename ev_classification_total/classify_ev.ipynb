# EV classification: total property samples

This notebook reads the immutable artifacts produced by `run_classification.py`. It does not fit models or reopen the held-out test set. The source is `data_addition/train_property_samples_auxw_netload.csv`; quarantined feed-in/export features are excluded.

```python
from pathlib import Path
import json
import pandas as pd
from IPython.display import display

ROOT = Path.cwd()
if not (ROOT / 'output').exists():
    ROOT = ROOT / 'ev_classification_total'
OUT = ROOT / 'output'
manifest = json.loads((OUT / 'run_manifest.json').read_text())
freeze = json.loads((OUT / 'final_freeze.json').read_text())
leaderboard = pd.read_csv(OUT / 'leaderboard.csv')
test_metrics = json.loads((OUT / 'best' / 'metrics_test.json').read_text())
print('Source:', manifest['source']['path'])
print('SHA-256:', manifest['source']['sha256'])
print('Frozen champion:', freeze['champion'])
```
